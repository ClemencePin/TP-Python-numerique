```python
#2.1
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio,display
from scipy.io import wavfile

f = 440 #hertz
fe = 44100
temps = np.linspace(0,1,44100)
tab = np.sin(2*np.pi*f*temps)

display(Audio(tab,rate = fe))
plt.plot(temps,tab)
plt.show()
```

```python
#2.2
def sine(freq,duration,amplitude):
    temps = np.linspace(0,duration,fe)
    tab = amplitude*np.sin(2*np.pi*freq*temps)
    return display(Audio(tab,rate = fe))
sine(440,1,1)
```

```python
#2.3
def sine_linear(freq1, freq2,duration):
    temps = np.linspace(0,duration,fe)
    freq = np.linspace(freq1,freq2,fe)
    tab = np.sin(2*np.pi*freq*temps)
    return display(Audio(tab,rate = fe))
sine_linear(440,523,1)
```

```python
#3
def crescendo_sine(freq,duration,creordecre):
    temps = np.linspace(0,duration,fe)
    if creordecre == True:
        amp = np.linspace(1,10,fe)
    else : 
        amp = np.linspace(10,1,fe)
    tab = np.sin(2*np.pi*freq*temps)*amp
    return display(Audio(tab,rate = fe))
crescendo_sine(440,1,False)

```

```python
#6.2.2
def freq_from_name(name):
    freqla = 440
    notes = ["do","dod","re","red","mi","fa","fad","sol","sold","la","lad","si"]
    freq = np.empty((12))
    freq[9] = freqla
    for i in range(12):
        freq[i] = freqla*2**((i-9)/12)
    
    dic = {}
    for i in range(12):
        dic[notes[i]]=freq[i]
    return dic[name]

def verif():
    return freq_from_name("la")==440
verif()
```

```python
#7
def accord(freq1,freq2):
    temps = np.linspace(0,1,fe)
    tab1 = np.sin(2*np.pi*freq1*temps)
    tab2 = np.sin(2*np.pi*freq2*temps)
    return display(Audio(tab1+tab2,rate = fe))
accord(440,420)
    
```
